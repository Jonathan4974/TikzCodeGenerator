```python
from pathlib import Path

import pandas as pd
from datasets import Dataset, Features, Value
from huggingface_hub import HfApi


FILES = [
    "datikz-test.parquet",
    "datikz-train.parquet",
]

HF_TOKEN = "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"
REPO_ID = "loss-boss/tikz-dataset"
OUTPUT = Path("new_bench_set_part-00000.parquet")


def get_image_bytes(image):
    if image is None:
        raise ValueError("Bild fehlt")

    if isinstance(image, dict):
        if image.get("bytes") is not None:
            return bytes(image["bytes"])
        return Path(image["path"]).read_bytes()

    return bytes(image)


# Dateien überprüfen
for file in FILES:
    if not Path(file).exists():
        raise FileNotFoundError(f"Nicht gefunden: {Path(file).resolve()}")

# Nur benötigte Spalten laden – nicht die große PDF-Spalte
frames = []

for file in FILES:
    print(f"Lade {file} ...", flush=True)

    part = pd.read_parquet(
        file,
        columns=["code", "image", "origin", "date"],
    )

    print(f"  {len(part)} Zeilen geladen", flush=True)
    frames.append(part)

df = pd.concat(frames, ignore_index=True)
print(f"Insgesamt: {len(df)} Zeilen", flush=True)

# Filter
df["date"] = pd.to_datetime(
    df["date"],
    errors="coerce",
    utc=True,
)

df = df[
    df["origin"]
    .fillna("")
    .astype(str)
    .str.strip()
    .str.lower()
    .eq("arxiv")
    & df["date"].dt.year.eq(2026)
].reset_index(drop=True)

print(f"Nach Filterung: {len(df)} Zeilen", flush=True)

if df.empty:
    raise ValueError("Der Filter hat keine passenden Zeilen gefunden.")

# Bilder verarbeiten
print("Verarbeite Bilder ...", flush=True)

images = []

for i, image in enumerate(df["image"]):
    images.append(get_image_bytes(image))

    if (i + 1) % 500 == 0:
        print(f"  {i + 1}/{len(df)} Bilder", flush=True)

# Zieldatensatz erstellen
dataset = Dataset.from_dict(
    {
        "image_with_text": images,
        "code_with_text": df["code"].astype(str).tolist(),
        "vlm_description": [None] * len(df),
        "source": ["new-bench-set"] * len(df),
    },
    features=Features(
        {
            "image_with_text": Value("binary"),
            "code_with_text": Value("string"),
            "vlm_description": Value("string"),
            "source": Value("string"),
        }
    ),
)

print(f"Speichere {OUTPUT} ...", flush=True)
dataset.to_parquet(OUTPUT)

print(
    f"Gespeichert: {OUTPUT.resolve()} "
    f"({OUTPUT.stat().st_size / 1024**2:.1f} MB)",
    flush=True,
)

print("Starte Upload ...", flush=True)

result = HfApi(token=HF_TOKEN).upload_file(
    path_or_fileobj=OUTPUT,
    path_in_repo=OUTPUT.name,
    repo_id=REPO_ID,
    repo_type="dataset",
    commit_message="Add new benchmark set",
)

print(f"Upload abgeschlossen: {result}", flush=True)
```
